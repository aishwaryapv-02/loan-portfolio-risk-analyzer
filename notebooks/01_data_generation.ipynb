```python
import pandas as pd
import numpy as np
import random
from faker import Faker
from datetime import datetime, timedelta
```

```python
random.seed(42)
Faker.seed(1000)
fake = Faker()
```

```python
customer_id = [fake.uuid4() for _ in range(1000)]
name = [fake.name() for _ in range(1000)]
age = [random.randint(18,99) for _ in range(1000)]
state = [fake.state() for _ in range(1000)]

customer_income = ['<30k','30k-60k','60k-90k','90k+']
income_bracket = random.choices(customer_income, k=1000)

credit = ['Poor','Fair','Good','Very Good','Excellent']
w = [0.05,0.15,0.3,0.3,0.2]
credit_score_band = random.choices(credit, weights = w, k=1000)
```

```python
customers = pd.DataFrame({'customer_id':customer_id,'name':name,'age':age,'state':state,
                          'income_bracket':income_bracket,'credit_score_band':credit_score_band})
customers
```

```python
customers['credit_score_band'].value_counts()
```

```python
cust_id = []
options = [1,2,3]
weight = [0.80,0.17,0.03]
vals = random.choices(options, weights = weight, k = 1000)

for cus,n in zip(customer_id,vals):
    for i in range(n):
        cust_id.append(cus)
```

```python
loan = pd.DataFrame({'customer_id':cust_id})
```

```python
l_id = []

for i in range(1001, len(cust_id)+1001):
    l_id.append(i)
    
len(l_id)
loan_id = ['L' + str(l) for l in l_id]
len(loan_id)
```

```python
loan = pd.DataFrame({'customer_id':cust_id, 'loan_id':loan_id})
loan
```

```python
band_score = {'Poor':(300,579),'Fair':(580,669),'Good':(670,739),'Very Good':(740,799),'Excellent':(800,850)}

customers['score'] = customers.apply(lambda row: random.randint(band_score[row['credit_score_band']][0],
                                                                band_score[row['credit_score_band']][1]), axis = 1)

customers
```

```python
customers['score'].describe()
```

```python
customers.groupby('credit_score_band')['score'].agg(['min','max'])
```

```python
def score_grade(row):
    if row['score'] >= 760:
        return 'A'
    elif row['score'] >= 720:
        return 'B'
    elif row['score'] >= 690:
        return 'C'
    elif row['score'] >= 660:
        return 'D'
    elif row['score'] >= 630:
        return 'E'
    elif row['score'] >= 600:
        return 'F'
    else:
        return 'G'

customers['grade'] = customers.apply(score_grade, axis = 1)

customers
```

```python
customers['grade'].value_counts()
```

```python
loans = pd.merge(loan, customers, on = 'customer_id', how = 'left')
```

```python
loans
```

```python
loans['term_months'] = 36
loans
```

```python
start = datetime(2023,1,1)
end = datetime(2024,12,31)

loans['origination_date'] = pd.to_datetime(loans.apply(lambda x:fake.date_between_dates(date_start = start, date_end = end), axis = 1))
loans
```

```python
loans.dtypes
```

```python
apr_dict = {'A':0.075,'B':0.105,'C':0.139,'D':0.178,'E':0.221,'F':0.262,'G':0.298}

loans['apr'] = loans.apply(lambda x: apr_dict[x['grade']], axis = 1)

loans
```

```python
loans['loan_amount'] = loans.apply(lambda x: round(random.randint(1000,40000)/500)*500, axis = 1)
loans
```

```python
(loans['loan_amount'] % 500 ==0).all()
```

```python
loans['scheduled_payments'] = round(loans['loan_amount'] * (loans['apr']/12) / (1-(1+(loans['apr']/12))**(-loans['term_months'])),2)
loans
```

```python
loans['vintage_month'] = loans['origination_date'].dt.to_period('M').dt.to_timestamp()
loans
```
